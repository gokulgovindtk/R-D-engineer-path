```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

k = 100
c = 0.5
m = 1.0   # mass

def dsdt(s, t):
    x, omega = s
    dxdt = omega
    domegadt = (-k*x - c*omega) / m
    return [dxdt, domegadt]

x0 = 1
omega0 = 0
s0 = (x0, omega0)

t = np.linspace(0, 20, 100)

sol = odeint(dsdt, s0, t)

x = sol[:, 0]
omega = sol[:, 1]

plt.plot(t, x)
plt.xlabel("Time")
plt.ylabel("Displacement")
plt.title("Damped Harmonic Oscillator")
plt.show()
```
