```python
import numpy as np

matrix_a = np.array([
    [1, 3, 5],
    [2, 6, 10],
    [3, 35, 23]
])
          
x=np.array([[1],
 [2],
 [9]])
c=matrix_a@x
print(c)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```
