```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

# time
t = np.linspace(0, 50, 100)

# population
x0 = 1.4 * 10**9
def pop(x0, t):
    r = 0.012
    p = x0 * np.exp(r * t)
    return p

z = pop(x0, t)

# consumption
def cons(z):
    a = 0.8
    b = 1.1
    n = np.random.normal(0, 0.05 * np.mean(z), size=len(z))
    c = a * z**b + n
    return c

k = cons(z)

# plotting
plt.figure(figsize=(8, 6))

plt.subplot(2, 2, 1)
plt.plot(t, z, label='population')
plt.xlabel("years")
plt.ylabel("population")
plt.title("population vs years")
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(t, k, label='consumption')
plt.xlabel("years")
plt.ylabel("consumption")
plt.title("consumption vs years")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(z, k, label='consumption vs population')
plt.xlabel("population")
plt.ylabel("consumption")
plt.title("consumption vs population")
plt.legend()

plt.tight_layout()
plt.show()

# curve fitting
def opt(x, a, b):
    return a * x**2 + b

popt, pcov = curve_fit(opt, k, z, p0=(1, 1))
popt

```
