```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
x_1=np.linspace(0,5,10)
y_1=x_1**2

plt.title('Days sqaured')
plt.xlabel('days')
plt.ylabel('days squared')
plt.subplot(1,2,1)
plt.plot(x_1,y_1,'r')
plt.subplot(1,2,2)
plt.plot(x_1,y_1,'b')
plt.tight_layout()
plt.plot(x_1,y_1)
```

```python
fig_1=plt.figure(figsize=(5,4),dpi=100)
axes_1=fig_1.add_axes([0.1,0.1,0.9,0.9])
axes_1.set_xlabel("Days")
axes_1.set_ylabel("Days squared")
axes_1.plot(x_1,y_1,label='x/x^2')
axes_1.plot(y_1,x_1,label='x^2/x')
axes_1.legend(loc=0)
axes_2=fig_1.add_axes([0.45,0.45,0.4,0.3])
axes_2.set_xlabel("Days")
axes_2.set_ylabel("Days squared")
axes_2.plot(x_1,y_1,label='x/x^2')
axes_2.plot(y_1,x_1,label='x^2/x')
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,np.pi,100)
y_1=np.sin(x)
y_2=np.exp(-x)
fig_1=plt.figure(figsize=(5,4),dpi=100)
axes_1=fig_1.add_axes([0.1,0.1,0.9,0.9])
axes_1.plot(x,y_1,'r',label='y = sin(x)')
axes_1.plot(x,y_2,'b',label='y = e^(-x)')


axes_1.set_xlabel('Radian')
axes_1.set_ylabel('value')
axes_1.set_title('sine function and exponential')

axes_1.legend() 


```
